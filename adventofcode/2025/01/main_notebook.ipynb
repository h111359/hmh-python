```python
def load_input(file_name):
    l_input = []
    with open(file_name, 'r') as file:
        for row in file:
            l_input.append(row.strip())
    return l_input
```

```python
def solve_part_01(l_input):
    
    position = 50
    finish_on_zero = 0
    
    for i in l_input:

        previous_position = position
        
        if i[0] == 'L':
            sign = -1
        elif i[0] == 'R':
            sign =  1
        else:
            continue

        position = previous_position + sign * int(i[1:])

        if position % 100 == 0:
           finish_on_zero += 1

    return(finish_on_zero)
        
```

```python
def solve_part_02(l_input):
    
    position = 50
    step_on_zero = 0
    
    for i in l_input:

        previous_position = position
        
        if i[0] == 'L':
            sign = -1
        elif i[0] == 'R':
            sign =  1
        else:
            continue

        distance = int(i[1:])
        position = previous_position + sign * distance        

        #begin_interval = previous_position if sign == 1 else position

        for p in range(previous_position + sign, position + sign, sign):
            if p % 100 == 0:
                step_on_zero += 1
                # print(f'Add to step_on_zero with p = {p} whitin the interval {previous_position + sign} .. {position + sign}')

        # if position % 100 == 0:
        #   step_on_zero += 1

        # print(f'previous_position = {previous_position}. i = {i}, position = {position}, step_on_zero = {step_on_zero}')

    return(step_on_zero)
        
```

```python
file_name = 'input.txt'
l_input = load_input(file_name)
result_01 = solve_part_01(l_input)
print(result_01)
result_02 = solve_part_02(l_input)
print(result_02)
```
