```python
import re
```

```python
def regex_find_pattern(s):
    l1 = re.findall('mul\(\w+,\w+\)',s)
    return l1
```

```python
summed = 0
with open('input.txt', 'r') as file:
    for line in file:
        mul_l = regex_find_pattern(line)
        list_1 = [x.lstrip('mul(') for x in mul_l]
        list_2 = [x.rstrip(')') for x in list_1]
        list_3 = [x.split(',') for x in list_2]
        for i in range(len(list_3)):
            (a, b) = list_3[i]
            #print('a={} b={}'.format(a, b))
            summed += int(a) * int(b)

print(summed)
```

```python
def regex_do(s):
    l1 = re.findall(r'do\(\)(.*?)don\'t\(\)', s)
    return l1
```

```python
def reduce_list_do(s):
    temp_s = s
    first_dont = temp_s.find('don\'t')

    if first_dont == -1:
        return temp_s
    else:
        a = temp_s[first_dont+7:]
        #print(a)
        enabler = a.find('do()')
        #print(enabler)
        if enabler == -1:
            return temp_s[:first_dont]
        else:
            return temp_s[:first_dont] + "-------------\n" + reduce_list_do(a[enabler+4:])
```

```python
def line_calc(line):
    summed = 0
    reduced_line = reduce_list_do(line)
    print(reduced_line)
    
    #list_1 = regex_do(reduced_line)
    
    #print(list_1)
    
    str_1 = "".join(reduced_line)
    
    mul_l = regex_find_pattern(str_1)
    list_1 = [x.lstrip('mul(') for x in mul_l]
    list_2 = [x.rstrip(')') for x in list_1]
    list_3 = [x.split(',') for x in list_2]
    for i in range(len(list_3)):
        (a, b) = list_3[i]
        #print('a={} b={}'.format(a, b))
        summed += int(a) * int(b)
    return summed
```

```python
print(line_calc("xmul(2,4)&mul[3,7]!^don't()_mul(5,5)+mul(32,64](mul(11,8)undo()?mul(8,5))"))
```
