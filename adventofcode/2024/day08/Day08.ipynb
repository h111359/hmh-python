```python
l_initial = []

with open('input.txt', 'r') as file:a
    for row in file:
        l_initial.append(list(row.strip()))

rows = len(l_initial)
cols = len(l_initial[0])
```

```python
def print_list(l):
    for i in range(len(l)):
        print(''.join(l[i]))
```

```python
def get_antenas(p_list):
    v_antenas = []
    for i in range(len(p_list)):
        for j in range(len(p_list[0])):
            if p_list[i][j] != '.':
                v_antenas.append((p_list[i][j], int(i), int(j)))
    return v_antenas
```

```python
l_antenas = get_antenas(l_initial)
```

```python
def calc_pair_antinodes(p_pair):
    (x1, y1) = (p_pair[0][1], p_pair[0][2])
    (x2, y2) = (p_pair[1][1], p_pair[1][2])

    row_dist = x2 - x1
    col_dist = y2 - y1

    return([(x1 - row_dist, y1 - col_dist),(x2 + row_dist, y2 + col_dist)])
```

```python
def get_antena_pairs(p_antenas):
    v_antena_pairs = []
    for i in range(len(p_antenas)):
        for j in range(i+1, len(p_antenas)):
            if p_antenas[i][0] == p_antenas[j][0]:     
                v_pair = [p_antenas[i], p_antenas[j]]
                v_antinodes = calc_pair_antinodes([p_antenas[i], p_antenas[j]])
                v_pair.extend(v_antinodes)
                v_antena_pairs.append(v_pair)
    return v_antena_pairs
```

```python
l_pairs = get_antena_pairs(l_antenas)
```

```python
def is_on_map(p_tuple):
    if p_tuple[0] >= 0 and p_tuple[0] < rows and p_tuple[1] >= 0 and p_tuple[1] < cols:
        return True
    return False
```

```python
l_antinodes = []
for pair in l_pairs:
    if is_on_map(pair[2]):
        l_antinodes.append(pair[2])
    if is_on_map(pair[3]):
        l_antinodes.append(pair[3])

d_antinodes = dict([(x, 1) for x in l_antinodes])
print(len(d_antinodes), d_antinodes)
```

```python
def get_antinodes_2(p_antena_1, p_antena_2):
    row_dist = p_antena_2[0] - p_antena_1[0]
    col_dist = p_antena_2[1] - p_antena_1[1]

    l_antinodes = []

    v_loc = p_antena_1
    
    while is_on_map(v_loc):
        l_antinodes.append(v_loc)
        v_loc = (v_loc[0] + row_dist, v_loc[1] + col_dist)

    v_loc = (p_antena_1[0] - row_dist, p_antena_1[1] - col_dist)

    while is_on_map(v_loc):
        l_antinodes.append(v_loc)
        v_loc = (v_loc[0] - row_dist, v_loc[1] - col_dist)

    return(l_antinodes)
```
