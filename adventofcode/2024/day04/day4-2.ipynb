```python
matr = []
with open('input.txt', 'r') as file:
    for line in file:
        row = list(line.strip())
        matr.append(row)

print(matr)
```

```python
def check_x(i,j,rows,cols):
    if matr[i][j] == 'A' and j-1>=0 and j+1<=cols and i-1>=0 and i+1<=rows:
        if (([matr[i-1][j+1], matr[i+1][j-1]] == ['M','S']) or ([matr[i-1][j+1], matr[i+1][j-1]] == ['S','M'])) and (([matr[i+1][j+1], matr[i-1][j-1]] == ['M','S']) or ([matr[i+1][j+1], matr[i-1][j-1]] == ['S','M'])):
            #print(i,j, ([matr[i-1][j+1], matr[i][j], matr[i+1][j-1]]), ([matr[i+1][j-1], matr[i][j], matr[i-1][j+1]]))
            return 1            
        else:
            return 0
    else:
        return 0
```

```python
sum = 0
cols = len(matr[0]) - 1
rows = len(matr) - 1
for i in range(cols+1):
    for j in range(rows+1):
        #print('i={} j={} rows={} cols={}'.format(i,j,rows,cols))
        sum += check_x(i,j,rows, cols)
print(sum)
```
