```python
matr = []
with open('input.txt', 'r') as file:
    for line in file:
        row = list(line.strip())
        matr.append(row)

print(matr)
```

```python
def check_right(i,j,rows,cols):    
    if matr[i][j] == 'X' and j+3<=cols:
        #print([matr[i][j], matr[i][j+1], matr[i][j+2], matr[i][j+3]])
        if matr[i][j+1] == 'M':
            if matr[i][j+2] == 'A':
                if matr[i][j+3] == 'S':
                    return 1
                else:
                    return 0
            else:
                return 0    
        else:
            return 0
    else:
        return 0
```

```python
def check_left(i,j,rows,cols):
    if matr[i][j] == 'X' and j-3>=0:
        if matr[i][j-1] == 'M':
            if matr[i][j-2] == 'A':
                if matr[i][j-3] == 'S':
                    return 1
                else:
                    return 0
            else:
                return 0    
        else:
            return 0
    else:
        return 0
```

```python
def check_down(i,j,rows,cols):
    if matr[i][j] == 'X' and i+3<=rows:
        if matr[i+1][j] == 'M':
            if matr[i+2][j] == 'A':
                if matr[i+3][j] == 'S':
                    return 1
                else:
                    return 0
            else:
                return 0    
        else:
            return 0
    else:
        return 0
```

```python
def check_up(i,j,rows,cols):
    if matr[i][j] == 'X' and i-3>=0:
        if matr[i-1][j] == 'M':
            if matr[i-2][j] == 'A':
                if matr[i-3][j] == 'S':
                    return 1
                else:
                    return 0
            else:
                return 0    
        else:
            return 0
    else:
        return 0
```

```python
def check_ru(i,j,rows,cols):
    if matr[i][j] == 'X' and j+3<=cols and i-3>=0:
        if matr[i-1][j+1] == 'M':
            if matr[i-2][j+2] == 'A':
                if matr[i-3][j+3] == 'S':
                    return 1
                else:
                    return 0
            else:
                return 0    
        else:
            return 0
    else:
        return 0
```

```python
def check_lu(i,j,rows,cols):
    if matr[i][j] == 'X' and j-3>=0 and i-3>=0:
        if matr[i-1][j-1] == 'M':
            if matr[i-2][j-2] == 'A':
                if matr[i-3][j-3] == 'S':
                    return 1
                else:
                    return 0
            else:
                return 0    
        else:
            return 0
    else:
        return 0
```

```python
def check_rd(i,j,rows,cols):
    if matr[i][j] == 'X' and j+3<=cols and i+3<=rows:
        if matr[i+1][j+1] == 'M':
            if matr[i+2][j+2] == 'A':
                if matr[i+3][j+3] == 'S':
                    return 1
                else:
                    return 0
            else:
                return 0    
        else:
            return 0
    else:
        return 0
```

```python
def check_ld(i,j,rows,cols):
    if matr[i][j] == 'X' and j-3>=0 and i+3<=rows:
        if matr[i+1][j-1] == 'M':
            if matr[i+2][j-2] == 'A':
                if matr[i+3][j-3] == 'S':
                    return 1
                else:
                    return 0
            else:
                return 0    
        else:
            return 0
    else:
        return 0
```

```python
sum = 0
cols = len(matr[0]) - 1
rows = len(matr) - 1
for i in range(cols+1):
    for j in range(rows+1):
        #print('i={} j={} rows={} cols={}'.format(i,j,rows,cols))
        sum += check_right(i,j,rows, cols)
        sum += check_left(i,j,rows, cols)
        sum += check_up(i,j,rows, cols)
        sum += check_down(i,j,rows, cols)
        sum += check_ru(i,j,rows, cols)
        sum += check_lu(i,j,rows, cols)
        sum += check_rd(i,j,rows, cols)
        sum += check_ld(i,j,rows, cols)

print(sum)
```
