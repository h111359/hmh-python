```python
def get_input(file_name):
    v_schematics = []
    with open(file_name, 'r') as file:
        sch = []
        for row in file:
            if row == '\n':
                v_schematics.append(sch)
                sch = []
            else:
                sch_row = list(row.strip())
                sch.append(sch_row)
        v_schematics.append(sch)
    return v_schematics

# # Test
# get_input('input.txt')
```

```python
def get_lock_vector(p_sch):
    v0, v1, v2, v3, v4 = (0, 0, 0, 0, 0)
    for row in p_sch[1:-1]:
        [p0, p1, p2, p3, p4] = row
        if p0 == '#': v0 += 1
        if p1 == '#': v1 += 1 
        if p2 == '#': v2 += 1 
        if p3 == '#': v3 += 1 
        if p4 == '#': v4 += 1 
    return [v0, v1, v2, v3, v4]

# # Test
# sch = [['#', '#', '#', '#', '#'],
#   ['#', '#', '#', '#', '#'],
#   ['#', '#', '.', '#', '#'],
#   ['#', '.', '.', '#', '.'],
#   ['.', '.', '.', '.', '.'],
#   ['.', '.', '.', '.', '.'],
#   ['.', '.', '.', '.', '.']]

# get_lock_vector(sch)
```

```python
def get_locks(p_input):
    l_locks = []
    for sch in p_input:
        if sch[0][0] == '#':
            lock_vector = get_lock_vector(sch)
            l_locks.append(lock_vector)
    return l_locks

# # Test
# get_locks(list_input)
```

```python
def get_key_vector(p_sch):
    v0, v1, v2, v3, v4 = (5, 5, 5, 5, 5)
    for row in p_sch[1:-1]:
        [p0, p1, p2, p3, p4] = row
        if p0 == '.': v0 -= 1
        if p1 == '.': v1 -= 1 
        if p2 == '.': v2 -= 1 
        if p3 == '.': v3 -= 1 
        if p4 == '.': v4 -= 1 
    return [v0, v1, v2, v3, v4]

# # Test
# sch = [['.', '.', '.', '.', '.'],
#   ['.', '.', '#', '.', '.'],
#   ['.', '.', '#', '#', '.'],
#   ['#', '.', '#', '#', '.'],
#   ['#', '.', '#', '#', '#'],
#   ['#', '.', '#', '#', '#'],
#   ['#', '#', '#', '#', '#']]

# get_key_vector(sch)
```

```python
def get_keys(p_input):
    l_keys = []
    for sch in p_input:
        if sch[0][0] == '.':
            key_vector = get_key_vector(sch)
            l_keys.append(key_vector)
    return l_keys

# # Test
# get_keys(list_input)
```

```python
def key_match_lock(k, l):
    for i in range(len(k)):
        if k[i] + l[i] > 5:
            return False
    return True
```

```python
list_input = get_input('input.txt')
list_locks = get_locks(list_input)
list_keys = get_keys(list_input)

# list_locks = [[0,5,3,4,3],[1,2,0,5,3]]
# list_keys = [[5,0,2,1,3], [4,3,4,0,2], [3,0,2,0,1]]

list_pairs = []

for l in list_locks:
    for k in list_keys:
        if key_match_lock(k, l):
            list_pairs.append([k, l])

# f"{''.join(key)}_{''.join(lock)}"
string_pairs = set([(''.join(str(x) for x in key) + '-' + ''.join(str(x) for x in lock)) for [key, lock] in list_pairs])

# Test
print(f'Locks: {len(list_locks)}')
print(f'Keys: {len(list_keys)}')
print(f'Len of pairs: {len(list_pairs)}')
print(f'Unuque pairs: {len(string_pairs)}')
```
