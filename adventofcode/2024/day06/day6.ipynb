```python
def is_on_map(p_map, pos):
    if pos[0] < 0 or pos[0] >= len(p_map) or pos[1] < 0 or pos[1] >= len(p_map[0]):
        return False
    return True
```

```python
def move(p_map, pos, debug):

    dr = {
        '^': (-1,0,'>'),
        '>': (0,1,'v'),
        'v': (1,0,'<'),
        '<': (0,-1,'^')
    }

    # Find out what is the current position direction
    direction = p_map[pos[0]][pos[1]]

    if debug:
        print('Processing position {} with direction {}'.format(pos, direction))

    # Based on the current position direction, what is the next step vector 
    #and what new direction shoeul be if we need to change the direction
    (rows_change, cols_change, next_direction) = dr[direction]

    # Calcuate the next position
    new_position = (pos[0] + rows_change, pos[1] + cols_change)
    if debug:
        print('The next position is {}'.format(new_position))

    # If the next position is outside the map - stop
    if not is_on_map(p_map, new_position):
        if debug:
            print('This is an exit position')
        return (new_position, True, False, direction) 
    
    if p_map[new_position[0]][new_position[1]] == '#':
        # If there is an obstacle on the next position, change direction but stay on the initial position
        new_position = pos
        direction = next_direction    
        if debug:
            print('There is an obstacle on the next. We stay in the same position {} and change its direction to {}'.format(new_position, direction))
    elif p_map[new_position[0]][new_position[1]] == direction:
        # Next position is on the field and there is no obstacle on it
        # and already has been passed with same direction - so it is a cicle
        if debug:
            print('We have been on the next position with the same direction. This is a cycle')
        return(pos, True, True, direction)
    
    p_map[new_position[0]][new_position[1]] = direction    
    if debug:
        print('Put in {} direction {} and go there'.format(new_position, p_map[new_position[0]][new_position[1]]))
    return (new_position, False, False, direction)

        
```

```python
def calculate_map(p_field_map, p_start_position, p_debug):
    is_end_position = False
    is_cycle = False
    v_position = p_start_position
    v_direction = ''
    v_passed_path = []
    while not is_end_position and not is_cycle:
        (v_position, is_end_position, is_cycle, v_direction) = move(p_field_map, v_position, p_debug)
        v_current_step = (v_position[0], v_position[1], v_direction)
        if v_current_step not in v_passed_path:            
            v_passed_path.append(v_current_step)
            #print(v_current_step)
        else:
            is_cycle = True
            if p_debug:
                print('\nCycle:\nCurrent step: {}\nCycle path: {}'.format(v_current_step, v_passed_path))
    return (v_position, is_end_position, is_cycle)
```
