```python
l_input = []
(max_rows, max_cols) = (-1, -1)
```

```python
with open('input.txt', 'r') as file:
    for row in file:
        l_input.append(list(row.strip()))
```

```python
def get_dimensions(l_input):
    return (len(l_input) - 1, len(l_input[0]) - 1)
```

```python
def gen_treadheads(p_input, p_row, p_col):

    current_hight = int(p_input[p_row][p_col])
    l_tread = [(p_row, p_col, int(p_input[p_row][p_col]))]

    #print('gen_treadheads: Execution for ({}, {}) with high {} and l_tread = {}'.format(p_row, p_col, current_hight, l_tread))
    
    if current_hight == 9:
        return l_tread
        
    if p_col + 1 <= max_cols and int(p_input[p_row][p_col + 1]) == current_hight + 1:
        l_tread.extend(gen_treadheads(p_input, p_row, p_col + 1))

    if p_col - 1 >= 0 and int(p_input[p_row][p_col - 1]) == current_hight + 1:
        l_tread.extend(gen_treadheads(p_input, p_row, p_col - 1))     

    if p_row - 1 >= 0 and int(p_input[p_row - 1][p_col]) == current_hight + 1:
        l_tread.extend(gen_treadheads(p_input, p_row - 1, p_col))       

    if p_row + 1 <= max_rows and int(p_input[p_row + 1][p_col]) == current_hight + 1:
        l_tread.extend(gen_treadheads(p_input, p_row + 1, p_col))  

    return l_tread
```

```python
def calculate_trailheads(p_input):
    for row in range(len(p_input)):
        for col in range(len(p_input[0])):
            #print('Calculating row = {}, col = {}, element = {}'.format(row, col, p_input[row][col]))
            if int(p_input[row][col]) == 0:
                trailheads.extend([gen_treadheads(p_input, row, col)])
                #print('Trail = {}\n'.format(trailheads))
```

```python
def reduce_trails(p_trails):
    l_reduced_trails = {}
    for tr in p_trails:
        initial_position = tr[0]
        s_end_points = set()
        high = 0
        for pos in tr:
            if pos[2] == 9:
                s_end_points.add(pos)
        l_reduced_trails[initial_position] = s_end_points
    return l_reduced_trails
```

```python
def calculate_sum_trailheads_score(p_reduced_trails):
    sum = 0
    for (start_pos, ends_set) in p_reduced_trails.items():
        #print('start_pos = {}, ends_set = {}'.format(start_pos, ends_set))
        sum += len(ends_set)
    return sum
```
