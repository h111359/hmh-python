```python
def get_input(file_name):

    initial_wires = dict()
    gate_wires = dict()
    
    with open(file_name, 'r') as file:
        row = file.readline()
        while row != '\n':
            wire, num = row.split(': ')
            initial_wires[wire.strip()] = int(num.strip())                    
            row = file.readline()

        for row in file:
            gate, wire = row.split(' -> ')
            gate_wires[wire.strip()] = gate 

    gates = dict([(gate, wire) for (wire, gate) in gate_wires.items()])
    x_wires = dict([(wire, val) for (wire, val) in initial_wires.items() if wire.startswith('x')])
    y_wires = dict([(wire, val) for (wire, val) in initial_wires.items() if wire.startswith('y')])    

    x_number = ''
    for i in sorted(x_wires.keys()):
        x_number = str(x_wires[i]) + x_number

    y_number = ''
    for i in sorted(y_wires.keys()):
        y_number = str(y_wires[i]) + y_number   

    expected_z_number = bin(int(x_number, 2) + int(y_number, 2)).lstrip('0b')

    return initial_wires, gate_wires, gates, x_wires, x_number, y_wires, y_number, expected_z_number


def decode_gate (p_gate):       
    left_wire, op, rigth_wire = p_gate.split()
    return left_wire, op, rigth_wire

def apply_op (p_left_wire, p_op, p_rigth_wire):
    if p_op == 'AND':
        res = p_left_wire & p_rigth_wire
    elif p_op == 'OR':
        res = p_left_wire | p_rigth_wire
    elif p_op == 'XOR':
        res = p_left_wire ^ p_rigth_wire
    return res

def calculate_wires (d_x_wires, d_y_wires, d_gates):
    d_wires = dict(d_x_wires) 
    d_wires.update(d_y_wires)
    all_wires_calculated = False

    while not all_wires_calculated:
        all_wires_calculated = True
        for gate, wire in d_gates.items():
            left_wire, op, rigth_wire = decode_gate(gate)
            if (wire not in d_wires) and (left_wire in d_wires) and (rigth_wire in d_wires):
                all_wires_calculated = False
                d_wires[wire] = apply_op(int(d_wires[left_wire]), op, int(d_wires[rigth_wire]))

    return d_wires

def calculate_z_number (d_wires):
    l_digits = [str(d_wires[z]) for z in sorted(d_wires, reverse=True) if z.startswith('z')]
    return ''.join(l_digits)


def calc_state(x_wires, y_wires, gates, expected_z_number):
    d_wires = calculate_wires(x_wires, y_wires, gates)
    z_actual = calculate_z_number (d_wires)
    v_diff = int(z_actual,2) - int(expected_z_number,2)
    wrong_wires = find_wrong_z_wires(expected_z_number, z_actual)

    return (d_wires, z_actual, v_diff, wrong_wires)

def find_wrong_z_wires(p_correct, p_actual):
    l_wrong_positions = []

    # if len(p_actual) < len(p_correct):
    #     l = len(p_correct)
    #     p_actual = f"{p_actual:0>{l}}"
    #     print(p_actual)
    
    for pos in range(-1, -1*len(p_actual)-1, -1):
        #pos = len(p_actual) - 1 - i

        #print(f"p_correct[{pos}] =  {p_correct[pos]}, p_actual[{pos}] = {p_actual[pos]}")
        
        try:
            if p_correct[pos] != p_actual[pos]:
                l_wrong_positions.append((-1 * pos) - 1)
        except:
            print(f"i = {i}, pos = {pos}")

    l_wrong_wires = [f"z{x:02d}" for x in l_wrong_positions]

    return l_wrong_wires

```

```python
def find_gate_key(d, side1, op, side2=''):

    if side2 != '':
        key = side1 + ' ' + op + ' ' + side2
        if key in d:
            return key
        else:
            key = side2 + ' ' + op + ' ' + side1
            if key in d:
                return key
            else:
                return ''
    else:
        possible_keys = [k for k in d if k.find(side1) > -1 and k.find(op) > -1]
        if len(possible_keys) == 1:
            return possible_keys[0]
        else:
            return possible_keys

# Test

find_gate_key(gates, 'wkr', 'XOR')
```
