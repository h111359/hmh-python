```python
def calculate_gate(p_wire, p_gate, p_calculated_wires, debug=False):
    wire1, operation, wire2 = p_gate.split(' ')
    print(f'wire1 = "{wire1}", operation = "{operation}", wire2 = "{wire2}"') if debug else None
    if (wire1 in p_calculated_wires) and (wire2 in p_calculated_wires):
        if operation == 'AND':
            p_calculated_wires[p_wire] = int(p_calculated_wires[wire1]) & int(p_calculated_wires[wire2])
        elif operation == 'OR':
            p_calculated_wires[p_wire] = int(p_calculated_wires[wire1]) | int(p_calculated_wires[wire2])
        elif operation == 'XOR':
            p_calculated_wires[p_wire] = int(p_calculated_wires[wire1]) ^ int(p_calculated_wires[wire2])
        return True
    return False
    
# # Test
# calculated_wires = dict(wires_i)
# calculate_gate('z00', 'x00 AND y00', calculated_wires, True)
# print(calculated_wires)
```

```python
def decode_gate (p_gate):       
    left_wire, op, rigth_wire = p_gate.split()
    return left_wire, op, rigth_wire
```

```python
def apply_op (p_left_wire, p_op, p_rigth_wire):
    if p_op == 'AND':
        res = p_left_wire & p_rigth_wire
    elif p_op == 'OR':
        res = p_left_wire | p_rigth_wire
    elif p_op == 'XOR':
        res = p_left_wire ^ p_rigth_wire
    return res
```

```python
def swap_gates(gates, gate1, gate2):
    w1 = gates[gate1]
    w2 = gates[gate2]
    new_gates = dict(gates)
    new_gates[gate1] = w2
    new_gates[gate2] = w1
    return new_gates, w1, w2
```
