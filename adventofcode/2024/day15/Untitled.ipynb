```python
import re
```

```python
class Warehouse:

    def read_file(self, p_file_name):
        l_field = []
        l_moves = []
        with open(p_file_name) as file:
            for row in file:
                
                if re.search("\A#", row):
                    l_field.append(list(row.strip()))

                if re.search("\A(<|>|\^|v)", row):
                    l_moves.extend(list(row.strip()))
        
        return (l_field, l_moves)
    
    def __init__(self, p_file_name):
        (self.field, self.moves) = self.read_file(p_file_name)
        self.robot_coordinates = self.find_robot()
        #self.stage_field = deepcopy(self.field)
        #print(self.robot_coordinates)self.stage_field

    def __str__(self):
        v_str = ''        
        
        for r in range(len(self.field)):
            v_str = v_str + f'r={r}: ' + ''.join(self.field[r]) + '\n'

        #v_str += '\n\n'        
        #v_str +=  ''.join(self.moves)

        v_str += '\n' 
        v_str += f'Robot: ({self.robot_coordinates[0]}, {self.robot_coordinates[1]})\n'
        
        return v_str.strip()


    def find_robot(self):
        for row in range(len(self.field)):
            for col in range(len(self.field[0])):
                #print(f'Looking for @ in ({row},{col}) = {self.field[row][col]}')
                if self.field[row][col] == '@':
                    return (row, col)

    def move_coordinates(self, p_coordinates, p_movement):
        return (p_coordinates[0] + p_movement[0], p_coordinates[1] + p_movement[1])

    def move_this(self, p_coordinates, new_coordinates):        
        if self.field[p_coordinates[0]][p_coordinates[1]] == '@':
            self.robot_coordinates = new_coordinates  
        self.field[new_coordinates[0]][new_coordinates[1]] = self.field[p_coordinates[0]][p_coordinates[1]]
        self.field[p_coordinates[0]][p_coordinates[1]] = '.'


    def gen_movement(self, p_direction):
        if p_direction == '<':
            movement = (0, -1)
        elif p_direction == '>':
            movement = (0, 1)
        elif p_direction == '^':
            movement = (-1, 0)
        elif p_direction == 'v':
            movement = (1, 0)

        return movement
   
    def move_object(self, p_object, p_coordinates, p_direction, recursion_depth, simulate = False):  

        movement = self.gen_movement(p_direction)
        new_coordinates = self.move_coordinates(p_coordinates, movement)
        #object_on_new_coordinates = self.field[new_coordinates[0]][new_coordinates[1]]
        
        if p_object == '.':
            return True

        if p_object == '#':
            return False

        if p_object == '@':
            submove = self.move_object(self.field[new_coordinates[0]][new_coordinates[1]], new_coordinates, p_direction, recursion_depth + 1, simulate)           
            if submove:
                if not simulate:
                    self.move_this(p_coordinates, new_coordinates)         
                return True
            else:
                return False
        
        if p_object == '[':
            sibling_coordinates = self.move_coordinates(p_coordinates, (0, 1))
            #sibling_object = self.field[sibling_coordinates[0]][sibling_coordinates[1]]
            #if sibling_object != ']':
            #    sibling_object = None
        elif p_object == ']':
            sibling_coordinates = self.move_coordinates(p_coordinates, (0, -1))
            #sibling_object = self.field[sibling_coordinates[0]][sibling_coordinates[1]]
            #if sibling_object != '[':
            #    sibling_object = None            

        if sibling_coordinates:
            sibing_new_coordinates = self.move_coordinates(sibling_coordinates, movement)

        if (recursion_depth == 0) or True:
            print('\n\nStart Move ------------------' )

            print(f'Move: simulation = {simulate} p_object = {p_object}, p_coordinates = {p_coordinates}, ' + 
                  f'new_coordinates = {new_coordinates}, object_on_new_coordinates = {self.field[new_coordinates[0]][new_coordinates[1]]}, ' + 
                  f'sibling_object = {self.field[sibling_coordinates[0]][sibling_coordinates[1]]}, sibling_coordinates = {sibling_coordinates}, ' +
                  f'p_direction = {p_direction}, recursion_depth = {recursion_depth}\n Over: \n')      
            
            print(self) 

            print('-' * 50)
            print('\n')

        if (p_object == '[' and p_direction == '>') or (p_object == ']' and p_direction == '<'):

            #submove = self.move_object(object_on_new_coordinates, new_coordinates, p_direction, recursion_depth + 1)
            submove1 = self.move_object(self.field[sibing_new_coordinates[0]][sibing_new_coordinates[1]], sibing_new_coordinates, p_direction, recursion_depth + 1, simulate)
            
            if submove1:
                if not simulate:
                    self.move_this(sibling_coordinates, sibing_new_coordinates)
                    self.move_this(p_coordinates, new_coordinates)                    
                return True
            else:
                return False

        if (p_object in ('[', ']')) and (p_direction in ('^', 'v')):
            submove = self.move_object(self.field[new_coordinates[0]][new_coordinates[1]], new_coordinates, p_direction, recursion_depth + 1, simulate)
            submove1 = self.move_object(self.field[sibing_new_coordinates[0]][sibing_new_coordinates[1]], sibing_new_coordinates, p_direction, recursion_depth + 1, simulate)
            
            if submove & submove1:
                if not simulate:
                    self.move_this(sibling_coordinates, sibing_new_coordinates)
                    self.move_this(p_coordinates, new_coordinates)                    
                return True
            else:
                return False        

            


    def execute_sequence(self):
        for m in self.moves:
            possible = self.move_object('@', self.robot_coordinates, m, 0, simulate = True)
            print('=' * 80)
            print('=' * 80)
            print(f'Move {self.robot_coordinates} to {m} with possibility = {possible}')
            if possible:
                self.move_object('@', self.robot_coordinates, m, 0, simulate = False)
            #print('\n\n')
            #print(self)

    def gps_coordinates_sum(self):
        v_sum = 0
        for row in range(len(self.field)):
            for col in range(len(self.field[0])):
                if self.field[row][col] == 'O':
                    v_sum += row*100 + col
        return v_sum

class Warehouse2(Warehouse):
    
    def __init__(self, p_warehouse):
        self.field = self.double_field(p_warehouse.field)
        self.moves = p_warehouse.moves
        #print(self.find_robot())
        self.robot_coordinates = self.find_robot()

    def double_field(self, p_field):
        l_field = []
        for r in p_field:
            l_row = []
            for c in r:
                if c == '#':
                    l_row.extend(['#','#'])
                elif c == '.':
                    l_row.extend(['.','.'])
                elif c == '@':
                    l_row.extend(['@','.'])
                elif c == 'O':
                    l_row.extend(['[',']'])
            l_field.append(l_row)
        return l_field

    def gps_coordinates_sum(self):
        v_sum = 0
        for row in range(len(self.field)):
            for col in range(len(self.field[0])):
                if self.field[row][col] == '[':
                    v_sum += row*100 + col
        return v_sum
```

```python
input_file = 'input.txt'

warehouse = Warehouse(input_file)

print(warehouse)

warehouse.execute_sequence()

print(warehouse)

print(warehouse.gps_coordinates_sum())
```
