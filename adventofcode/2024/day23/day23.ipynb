```python
def gen_tripples(p_set, l_input, debug=False):
    l_tripples = []
    l_comps = list(p_set)
    comp_cnt = len(l_comps)
    for c1 in range(comp_cnt - 2):
        print(f'calculating l_comps[{c1}] = {l_comps[c1]}')
        for c2 in range(c1, comp_cnt - 1):
            for c3 in range (c2, comp_cnt):
                if (((l_comps[c1], l_comps[c2]) in l_input) or ((l_comps[c2], l_comps[c1]) in l_input)) and\
                   (((l_comps[c2], l_comps[c3]) in l_input) or ((l_comps[c3], l_comps[c2]) in l_input)) and\
                   (((l_comps[c3], l_comps[c1]) in l_input) or ((l_comps[c1], l_comps[c3]) in l_input)) and\
                (l_comps[c1].startswith('t') or l_comps[c2].startswith('t') or l_comps[c3].startswith('t')):
                
                    l_tripples.append((l_comps[c1], l_comps[c2], l_comps[c3]))
    return l_tripples

# Test
print(l_input_sample, end='\n\n\n')
print(gen_tripples(s_comps_sample, l_input_sample))

# co,de,ta
# co,ka,ta
# de,ka,ta
# qp,td,wh
# tb,vc,wq
# tc,td,wh
# td,wh,yn
        
```

```python
def get_list_input(file_name):
    v_input = []
    with open(file_name, 'r') as file:
        for row in file:
            (a, b) = row.strip().split('-')
            v_input.append([a, b])
    return v_input
```
