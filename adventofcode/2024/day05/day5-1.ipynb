```python
l_order = []
l_pages = []
```

```python
mode = 'pairs'

with open('input.txt', 'r') as file:
    for l in file:
        if l == '\n':
            mode = 'pages'
        elif mode == 'pairs':
            l_order.append(l.strip().split('|'))
        else:
            l_pages.append(l.strip().split(','))
```

```python
def check_pair(a, b):
    for i in range(len(l_order)):
        if l_order[i] == [b, a]:
            return False
    return True
```

```python
def is_page_ordered(l):
    for i in range(len(l)):
        for j in range(i+1, len(l)):
            if not check_pair(l[i], l[j]):
                return 0
    return int(l[len(l)//2])
```

```python
s = 0
for l in l_pages:
    s += is_page_ordered(l)

print(s)
```

```python
s = 0
mid_page = 0
l_not_ordered_pages = []
for l in l_pages:
    mid_page = is_page_ordered(l)
    if mid_page == 0:
        l_not_ordered_pages.append(l)
```

```python
def sort_pages(p_list):
    swaps = 1
    l = p_list
    while swaps > 0:
        for i in range(len(l)):
            if i == 0:
                swaps = 0
            for j in range(i, len(l)):
                if not check_pair(l[i], l[j]): # Swap
                    #print(l[i], l[j])
                    (l[i], l[j]) = (l[j], l[i])
                    swaps += 1
    return l
```

```python
s = 0
for l in l_not_ordered_pages:
    l = sort_pages(l)

for l in l_not_ordered_pages:
    s += is_page_ordered(l)

print(s)
```
